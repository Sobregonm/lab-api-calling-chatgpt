```python
import json

def validate_listing(listing):
    description = listing.get("description", "")
    features = listing.get("features", [])
    keywords = listing.get("keywords", [])

    word_count = len(description.split())
    errors = []

    if not 150 <= word_count <= 200:
        errors.append(
            f"Description must have 150-200 words. Found: {word_count}"
        )

    if not 5 <= len(features) <= 7:
        errors.append("Features must contain 5-7 items")

    if not 10 <= len(keywords) <= 15:
        errors.append("Keywords must contain 10-15 items")

    return errors


# Test the validator without using the API
test_listing = {
    "description": "Example product description",
    "features": ["Feature 1", "Feature 2"],
    "keywords": ["Keyword 1"]
}

errors = validate_listing(test_listing)

print("\nVALIDATION RESULTS")
print("=" * 50)

if not errors:
    print("Product listing passed all validations!")
else:
    print("Product listing failed validation.")
    for error in errors:
        print("-", error)
```

```python
import json

# Load the previously generated product listing
with open("first_product_listing.json", "r", encoding="utf-8") as file:
    real_listing = json.load(file)

# Validate the real product
errors = validate_listing(real_listing)

print("\nREAL PRODUCT VALIDATION")
print("=" * 50)

if not errors:
    print("Product listing passed all validations!")
else:
    print("Product listing failed validation.")

    for error in errors:
        print("-", error)
```

```python
print("\nFINAL QUALITY REPORT")
print("=" * 50)

print("Product:", real_listing.get("title"))
print("Description words:", len(real_listing.get("description", "").split()))
print("Features:", len(real_listing.get("features", [])))
print("Keywords:", len(real_listing.get("keywords", [])))
print("Validation errors:", len(validate_listing(real_listing)))
```

```python
import json
from pathlib import Path

# Files to validate
listing_files = [
    Path("first_product_listing.json"),
    Path("generated_listings/product_39386.json"),
    Path("generated_listings/product_59263.json")
]

print("\nBATCH VALIDATION REPORT")
print("=" * 50)

for file_path in listing_files:

    with open(file_path, "r", encoding="utf-8") as file:
        product_listing = json.load(file)

    errors = validate_listing(product_listing)

    print("\nProduct:", product_listing.get("title"))
    print("Description words:",
          len(product_listing.get("description", "").split()))
    print("Features:", len(product_listing.get("features", [])))
    print("Keywords:", len(product_listing.get("keywords", [])))

    if not errors:
        print("Status: PASSED")
    else:
        print("Status: FAILED")
        for error in errors:
            print("-", error)

    print("-" * 50)
```

```python
# Identify products that need correction

failed_products = []

for file_path in listing_files:

    with open(file_path, "r", encoding="utf-8") as file:
        product_listing = json.load(file)

    errors = validate_listing(product_listing)

    if errors:
        failed_products.append({
            "file": str(file_path),
            "title": product_listing.get("title"),
            "errors": errors
        })

print("\nPRODUCTS REQUIRING CORRECTION")
print("=" * 50)

print("Total failed products:", len(failed_products))

for product in failed_products:
    print("\nProduct:", product["title"])
    print("File:", product["file"])

    for error in product["errors"]:
        print("Error:", error)
```
